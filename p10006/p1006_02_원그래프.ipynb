```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False    #마이너스(-) 부호 처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic' #한글처리 :윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15                #전체 글자폰트크기 조정
#matplotlib.rcParams['font.family'] = 'Apple Gothic' #맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
df=pd.read_excel('file/students.xlsx')
# df.groupby('학년').value_counts().loc['디지털고']
school = df['학교'].value_counts(normalize=True)
explode = [0.1,0,0]
school.index
school.values
plt.pie(school.values,labels=school.index,startangle=90,counterclock=False,explode=explode,
        autopct='%.1f%%')  
plt.show()
```

```python
# values=[1,1,1,1,1,1]
values=[30,25,20,13,10,2] #백분율로 자동계산
labels = ['파이썬','html','자바스크립트','판다스','그래프','디장고']
explode = [0.2,0.1,0,0,0,0]  #원파트 띄우기
colors = ['#ffadad','#ffd6a5','#fdffb6','#caffbf','#9bf6ff','#a0c4ff']
#startangle=기준각도,counterclock=시계방향
# autopct='%.1f%%'
plt.pie(values,labels=labels,startangle=90,counterclock=False,explode=explode,
        autopct='%.1f%%',colors=colors)  
plt.title('언어별 선호도')
plt.legend(loc=(1.2,0.3),title='언어별 선호도')
plt.show()
```

```python
#bear.spirit,wine => 합계 : total_servings
#순위 높은순 부터 정렬해서, 합계에 대한 막대 그래프를 출력하시오.
df['total_servings'] = df['beer_servings'] + df['spirit_servings'] + df['wine_servings']
```

```python
# 한국 막대그래프를 red 추가해서 출력하시ㅗㅇ.
sort_df = df.sort_values('total_servings',ascending=False)
sort_df['country']
sort_df['total_servings']
k_sort_df = sort_df[sort_df['country']=='South Korea']
x = sort_df['country']        #전체 국가이름
y = sort_df['total_servings'] #전체 국가별 합계
x2 = k_sort_df['country']  # 한국이름
y2 = k_sort_df['total_servings'] #한국합계

plt.bar(x,y,label='국가별')
plt.bar(x2,y2,color='r',label='한국') #한국그래프
plt.legend()
plt.xticks(x,idx)
plt.show()
```
