```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False   
matplotlib.rcParams['font.family'] = 'Malgun Gothic' 
matplotlib.rcParams['font.size'] = 15                
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/students.xlsx',index_col="번호")
df
```

```python
li = df.groupby('학교')[['국어','영어','수학']].mean()
li
```

```python
index = np.arange(li.shape[0])
index
```

```python
li = df.groupby('학교')[['국어','영어','수학']].mean().round(1)
index = np.arange(li.shape[0])

x = li.index

y = li['국어']
y2 =li['영어'] 
y3 = li['수학']

plt.figure(figsize=(20,10))

plt.title('학생성적평균그래프')
plt.bar(index-0.2,y,label='국어',width=0.2)
plt.bar(index,y2,label='영어',width=0.2)
plt.bar(index+0.2,y3,label='수학',width=0.2)

for i,(v,v2,v3) in enumerate(zip(y,y2,y3)):
    plt.text(i-0.2,v+2,v,ha='center',va='center')
    plt.text(i,v2+2,v2,ha='center',va='center')
    plt.text(i+0.2,v3+2,v3,ha='center',va='center')


plt.xticks(index, x)
plt.yticks([0,20,40,60,80,100])
plt.legend(ncol=3)
plt.show()
```

```python
#학교별, 학년별 국어,영어,수학 평균을 출력
df.groupby(['학교','학년'])[['국어','영어','수학']].mean()
```

```python
#합계,평균,등수 컬럼을 추가하고 값을 입력해서 출력하시오.
df['합계'] = df['국어'] + df['영어'] + df['수학']
df['평균'] = (df['합계'] / 3).round(2)
df['등수'] = df['합계'].rank(ascending=False).astype(int)
df
```
