```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False    #마이너스(-) 부호 처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic' #한글처리 :윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15                #전체 글자폰트크기 조정
#matplotlib.rcParams['font.family'] = 'Apple Gothic' #맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
# 선 스타일
x = [1,2,3]
y = [2,4,8]

plt.title('꺾은선 graph')  # 상단타이틀

#figure : 그래프 사이즈 변경 figuresize : 크기,dpi: 해상도,facecolor:배경색 변경
# plt.figure(figsize=(20,5))  #배경사이즈
# plt.figure(figsize=(10,5),dpi=200,facecolor='yellow')
plt.plot(x,y,linewidth=2,marker='o',ms=10,mfc='yellow',linestyle='-.',\
        color='blue')
#그래프 안에 글자 입력 : plt.text
# plt.text(x[0],y[0]+0.3,2,ha='center',color='red')  #ha='center' 중앙
for idx,value in enumerate(y):
    plt.text(x[idx],y[idx]+0.2,value,ha='center',color='red')

# X선 설명부분
# plt.xlabel('매출액',color='red',loc='right') # X축설명글 색상, 위치
# plt.ylabel('월별',color='#00aa00',loc='top') #Y축설명글 색상, 위치


# plt.xticks([1,1.2,1.4,1.6,2,3])
# plt.yticks([-15,10,20,30]) #간격부분 조정을 하면 꺾이는 상태를 조정
# # plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정
# plt.savefig('file/aaa2.png',dpi=200)  #스크린 샷 저장
plt.show()
```

```python
# 선 스타일
x = [1,2,3]
y = [20,40,80]
y2 = [50,10,30]
y3 = [10,20,50]

plt.title('꺾은선 graph')  # 상단타이틀
# plt.figure(figsize=(10,5),dpi=200,facecolor='yellow')
plt.plot(x,y,label='국어',linewidth=2,marker='o',ms=10,mfc='yellow',linestyle='-',\
        color='pink')
plt.plot(x,y2,label='영어',linewidth=2,marker='x',ms=10,mfc='yellow',linestyle=':',\
        color='blue')
plt.plot(x,y3,label='수학',linewidth=2,marker='^',ms=10,mfc='yellow',linestyle='-.',\
        color='green')
plt.legend()        #범례

#그래프 안에 글자 입력 : plt.text
# plt.text(x[0],y[0]+0.3,2,ha='center',color='red')  #ha='center' 중앙
for idx,value in enumerate(y):
    plt.text(x[idx],y[idx]+0.2,value,ha='center',color='red')
for idx,value in enumerate(y2):
    plt.text(x[idx],y2[idx]+0.2,value,ha='center',color='red')
for idx,value in enumerate(y3):
    plt.text(x[idx],y3[idx]+0.2,value,ha='center',color='red')


# plt.xticks([1,1.2,1.4,1.6,2,3])
# # plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정

plt.show()
```

```python
x = [1,2,3]
#x = ['홍길동','이순신','유관순']
y = [20,40,80]
y2 = [50,10,30]
y3 = [10,20,50]
colors = ['r','g','b']
plt.title('학생성적그래프')
#color:그래프 색상변경: 색상('red') 여러개-["red",'green'], width:막대그래프 넓이 조절, alpha:투명도
bar=plt.bar(x,y,color=colors,width=0.8,alpha=0.5)
bar[0].set_hatch('/')
bar[1].set_hatch('x')
bar[2].set_hatch('..')

# grid - 격자선, 안내선
plt.grid(axis='y',linewidth=2,color='purple',linestyle='-.',alpha=0.5)
plt.grid(axis='x')


#가로막대그래프 생성/ 가로는 height,새로는 width
#plt.barh(y,x,height=5)
#rotation=각도조절
# plt.xticks(x,rotation=45)
for i,v in enumerate(y):
    plt.text(x[i],y[i],v,ha='center') 
plt.show()
```
