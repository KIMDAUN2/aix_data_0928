```python
import pandas as pd
df= pd.read_csv('file/chipotle.tsv',sep='\t')
df
```

```python
df.shape
df['order_id'].unique()
df.columns
# df.info()
#df.describe()
```

```python
df['item_price'] = df['item_price'].str.replace('$', '').str.strip().astype(float)
df['order_id']= df['order_id'].astype(str)
```

```python
# 1. 가장 많이 주문한(order_id) item_name 10개를 출력하시오.
df['item_name'].value_counts().head(10)
```

```python
# 2. item_name 당 주문 총량(quantity) 순서로 출력하시오.
df.groupby('item_name')['quantity'].sum().sort_values(ascending=False)
```

```python
# 3. item_name 가장 많이 주문한 10개 출력하시오
df.groupby('item_name')['order_id'].count().sort_values(ascending=False).head(10)

#선생님 정답
# item_count.index[0]
#----------------------------
# item_count = df['item_name'].value_counts()[:10]
# item_count
```

```python
#합계(아이템 같은 이름 수량 총 합계)
df.groupby('item_name')['quantity'].sum()
```

```python
# 4. 주문당 평균 금액을 출력하시오.
df.groupby('order_id')['item_price'].sum().mean()
```

```python
#5. 한 주문에 50달러 이상 사용한 주문 번호(id) 출력
chipo = df.groupby('order_id').sum()
results = chipo[chipo.item_price >= 50]

results.index.values
results[:10]
```
